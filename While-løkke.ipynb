## Ekstra utfordring
Oppgaven under er fra en R2 eksamen, men med en lur while-løkke kan du som holder på å lære deg 1T løse oppgaven. Jeg løser oppgave a som inspirasjon , forklar de fire linjene i løkken før du prøver på resten.   



```python
from math import pi
# startverdier
D = 5 # diameter i cm
L_tot = pi*D #Total lengde av dopapir i cm 
n = 1 # antall runder med dopapir

# ruller på en og en runde på dorullen med en while løkke
while n<50:
    D = D + 0.03
    L = pi*D
    L_tot = L_tot + L
    n += 1
print(f'Når dorullen har {n} runder igjen er det {L_tot/100:.1f} m papir igjen')
```
