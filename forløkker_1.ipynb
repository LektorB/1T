# for - løkker
En av styrkene til datamaskiner er at de kan gjøre enkle operasjoner mange ganger etter hverandre. Vi lager et program som kjører rundt og rundt i en løkke (Loop på engelsk). En type løkke er *for - løkke* som vi skal undersøke i denne notebooken. Du kan også lese mer på side 419 i Mønster 1T boken


Vi lager en for-løkke ved hjelp av kode ordet `for` etterfulgt av et fritt valgt variabelnavn og kontrollordet `in`. Så følger en mengde som vi skal løpe gjennom og vi avslutter med et kolon. Koden som skal gjentas for hver runde i løkken må rykkes inn et hakk (skjer automatisk i de fleste kodeeditorer). 

Kjør og studer eksemplene under og gjør gjerne endringer i dem.


**Eksempel 1**   
Første eksempel løper gjennom en streng:

```python
streng = 'God morgen'

for bokstaver in streng: # for alle bokstavene som er i streng gjør:
    print(bokstaver)     # Det som skal gjøres
```

**Eksempel 2**    
Vi kan bruke for-løkker med lister:  
Du kan lese mer om lister på side 425 i mønster boken


```python
xverdier = [-5,-3,0,1,4]

for x in xverdier:
    print(x)
```

**Eksemple 3**   
Vi kan regne med variablen som vi definerer i løkken: 

```python
xverdier = [-5,-3,0,1,4]

for x in xverdier:
    print(x/2)
```

**Eksempel 4**  
Vi kan bruke løkker til å lage en liten verditabell

```python
xverdier = [-5,-3,0,1,4]

print(f'  x     y')    #lager overskrift
print('------------')
      
for x in xverdier:
    y = 2*x + 1          # regner ut en og en y-verdi
    print(f'{x}   {y}')  # Skriver ut en og en linje med x- og tilhørende y-verdier
    
```

**Eksemple 5**    
Tabellen vår ble litt rotete. Det er mulig å rette den opp ved å formatere f-strengen.    
**PS:** Dette er som regel unødvendig pynt og for avansert for oss i 1T, men det er mulig om du er ekstra interssert.


```python
xverdier = [-5,-3,0,1,4]

print(f'  x     y')
print('-----------')
      
for x in xverdier:
    y = 2*x + 1
    print(f'{x:>3}   {y:>3}') # : formater, > høyrestill, 3 bruk tre plasser til tallet.  
```
